```python
import pandas as pd
from evidently import Report
from evidently.presets import DataDriftPreset

reference = pd.read_parquet("../data/reference/churn_reference_v1.parquet")

current = reference.copy()
current["events_day1"] = current["events_day1"] * 1.8   # break ONE real column

report = Report([DataDriftPreset(method="psi")], include_tests=True)
my_eval = report.run(current, reference)     # current FIRST, reference second
my_eval                                       # renders the interactive report inline
```
